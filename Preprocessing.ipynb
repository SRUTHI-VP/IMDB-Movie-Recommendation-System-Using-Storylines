```python
import pandas as pd
import re

# Load your dataset
df = pd.read_csv("imdb_movies_2024(2).csv")

# Remove leading numbers + dot + space (e.g., "1. The Fall Guy" → "The Fall Guy")
df["Movie_Title"] = df["Movie_Title"].apply(lambda x: re.sub(r'^\d+\.\s*', '', str(x)))

# Save cleaned version
df.to_csv("imdb_movies_cleaned(2).csv", index=False)

print(df.head(20))
```
