```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

# Setup driver
driver = webdriver.Chrome()
driver.get("https://www.imdb.com/search/title/?title_type=feature&release_date=2024-01-01,2024-12-31&count=250")

# Scroll and click "250 more" multiple times
# Scroll and click "250 more" multiple times
body = driver.find_element(By.TAG_NAME, 'body')
for i in range(24):  # 24 × 250 = 6000 movies
    for j in range(120):
        body.send_keys(Keys.PAGE_DOWN)
    body.send_keys(Keys.PAGE_UP)
    time.sleep(5)
    try:
        load_more = driver.find_element(By.XPATH, "//span[contains(text(),'250 more')]")
        driver.execute_script("arguments[0].scrollIntoView(true);", load_more)
        driver.execute_script("arguments[0].click();", load_more)
        time.sleep(10)
    except:
        break  # no more button

# Collect movies
all_movies = {}
movie_name = driver.find_elements(By.XPATH, "//h3[contains(@class,'ipc-title__text')]")
movie_name_storyline = driver.find_elements(By.XPATH, "//div[contains(@class,'ipc-html-content-inner-div')]")

for movie, movie_story in zip(movie_name, movie_name_storyline):
    all_movies[movie.text] = movie_story.text

# Convert to DataFrame
df = pd.DataFrame(list(all_movies.items()), columns=["Movie_Title", "Storyline"])

# Print results
print(df.head(20))   # show first 20 rows
print(f"\nTotal movies scraped: {len(df)}")

# Optional: display full DataFrame in Jupyter
df
```

```python
df.to_csv("Imdb_movies_2024.csv", index=False, encoding="utf-8")
```
